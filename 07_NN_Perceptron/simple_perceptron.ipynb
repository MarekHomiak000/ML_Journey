```python
import numpy as np
```

```python
class Perceptron:
    def __init__(self, lr, weights, epochs):
        self.lr = lr
        self.weights = np.zeros(weights)
        self.epochs = epochs
        self.bias = 0.0

    def predict(self, X):
        z = np.dot(X, self.weights) + self.bias
        return self._activation_f(z)

    def _activation_f(self, z):
        return 1 if z >= 0 else 0

    def fit(self, X, y):
        for _ in range(self.epochs):
            for xi, yi in zip(X,y):
                loss = yi - self.predict(xi)
                self.weights += self.lr * loss * xi
                self.bias += self.lr * loss
```

```python
if __name__ == "__main__":
    X = np.array([[0,0],[0,1],[1,0],[1,1]])
    y = np.array([0,0,0,1])

    p = Perceptron(0.2, 2, 20)
    p.fit(X,y)

    for xi, yi in zip(X, y):
        print(f"{xi} → predikcia: {p.predict(xi)}, skutočnosť: {yi}")
```
